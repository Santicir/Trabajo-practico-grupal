```python
# 3. CARGA DE DATOS
# =============================================================================
def generar_datos_demo(n_partidos=1500, semilla=SEMILLA):
    """
    Datos SINTÉTICOS con estructura tipo FBref (formato 'ancho'), útiles para
    probar el pipeline. Las relaciones están simuladas: NO sacar conclusiones
    futbolísticas de este modo.
    """
    rng = np.random.default_rng(semilla)
    equipos = [f"Equipo_{i:02d}" for i in range(1, 21)]
    fuerza = dict(zip(equipos, rng.normal(0, 0.35, len(equipos))))

    def stats(xg, pos):
        return {
            "posesion": pos,
            "tiros_al_arco": rng.poisson(np.clip(0.4 + 2.6 * xg, 0.5, None)),
            "xg": xg,
            "pases_completados_pct": np.clip(77 + 0.15 * (pos - 50) + rng.normal(0, 4), 60, 95),
            "faltas": rng.poisson(np.clip(12.5 - 0.08 * (pos - 50), 5, None)),
            "tarjetas": None,  # se completa abajo (depende de faltas)
            "duelos_ganados": np.round(np.clip(rng.normal(50 - 0.10 * (pos - 50), 6), 25, 80)),
            "recuperaciones": rng.poisson(np.clip(46 - 0.25 * (pos - 50), 20, None)),
        }

    filas = []
    for i in range(n_partidos):
        loc, vis = rng.choice(equipos, 2, replace=False)
        d = fuerza[loc] - fuerza[vis]
        xg_l = float(np.clip(rng.gamma(6, 1.45 * np.exp(0.35 * d) / 6), 0.1, 5))
        xg_v = float(np.clip(rng.gamma(6, 1.15 * np.exp(-0.35 * d) / 6), 0.1, 5))
        pos_l = float(np.clip(50 + 8 * d + 5 * (xg_l - xg_v) + rng.normal(0, 7), 22, 78))
        s_l, s_v = stats(xg_l, pos_l), stats(xg_v, 100 - pos_l)
        for s in (s_l, s_v):
            s["tarjetas"] = rng.poisson(np.clip(0.8 + 0.10 * s["faltas"], 0.3, None))
        fila = {"partido_id": i, "local": loc, "visitante": vis,
                "goles_local": rng.poisson(xg_l), "goles_visitante": rng.poisson(xg_v)}
        fila.update({f"local_{k}": v for k, v in s_l.items()})
        fila.update({f"visitante_{k}": v for k, v in s_v.items()})
        filas.append(fila)

    df = pd.DataFrame(filas)
    # ~3 % de faltantes aleatorios para ejercitar la imputación
    cols_stats = [c for c in df.columns if c.startswith(("local_", "visitante_"))]
    mascara = rng.random((len(df), len(cols_stats))) < 0.03
    df[cols_stats] = df[cols_stats].mask(mascara)
    return df


def cargar_datos():
    """Carga el CSV según MODO_CARGA y devuelve el DataFrame crudo."""
    if MODO_CARGA == "demo":
        print("!" * 78)
        print("  MODO DEMO: datos SINTÉTICOS. Sirve para probar el código, NO para")
        print("  extraer conclusiones reales. Cambiá MODO_CARGA para usar tus datos.")
        print("!" * 78)
        return generar_datos_demo()

    if MODO_CARGA == "drive":
        if not en_colab():
            raise RuntimeError("El modo 'drive' solo funciona dentro de Google Colab.")
        from google.colab import drive
        drive.mount("/content/drive")
        ruta = RUTA_DRIVE_CSV
    elif MODO_CARGA == "upload":
        if not en_colab():
            raise RuntimeError("El modo 'upload' solo funciona dentro de Google Colab.")
        from google.colab import files
        print("Seleccioná tu archivo CSV de partidos...")
        subido = files.upload()
        ruta = next(iter(subido))
    elif MODO_CARGA == "local":
        ruta = RUTA_LOCAL_CSV
    else:
        raise ValueError("MODO_CARGA debe ser 'demo', 'upload', 'drive' o 'local'.")

    # sep=None -> detecta automáticamente coma, punto y coma o tabulación
    df = pd.read_csv(ruta, sep=None, engine="python", encoding="utf-8-sig")
    df.columns = [str(c).strip() for c in df.columns]
    df = df.rename(columns=RENOMBRAR_COLUMNAS)
    print(f"CSV cargado: {ruta}  ->  {df.shape[0]} filas x {df.shape[1]} columnas")
    return df

```

```python
# 4. PREPARACIÓN DE DATOS
# =============================================================================
def _a_numerico(serie):
    """Convierte a número tolerando '55%', '1,25' y textos basura."""
    if serie.dtype == object:
        serie = (serie.astype(str).str.replace("%", "", regex=False)
                 .str.replace(",", ".", regex=False).str.strip())
    return pd.to_numeric(serie, errors="coerce")


def _ancho_a_largo(df, features):
    """1 fila por partido -> 2 filas (perspectiva local y visitante)."""
    for c in ("goles_local", "goles_visitante"):
        if c not in df.columns:
            raise ValueError(f"Falta la columna obligatoria '{c}'.")
    df = df.copy()
    if "partido_id" not in df.columns:
        df["partido_id"] = np.arange(len(df))
    for c in ["goles_local", "goles_visitante"] + \
             [f"{l}_{f}" for f in features for l in ("local", "visitante")]:
        df[c] = _a_numerico(df[c])

    def lado(propio, rival, es_local):
        out = pd.DataFrame({
            "partido_id": df["partido_id"].values,
            "es_local": es_local,
            "goles_favor": df[f"goles_{propio}"].values,
            "goles_contra": df[f"goles_{rival}"].values,
        })
        for f in features:
            out[f] = df[f"{propio}_{f}"].values
            out[f"{f}__rival"] = df[f"{rival}_{f}"].values
        return out

    return pd.concat([lado("local", "visitante", 1),
                      lado("visitante", "local", 0)], ignore_index=True)


def _largo_con_rival(df, features):
    """1 fila por equipo-partido -> agrega las estadísticas del rival."""
    for c in ("partido_id", "equipo", "goles_favor", "goles_contra"):
        if c not in df.columns:
            raise ValueError(f"Falta la columna obligatoria '{c}' (formato largo).")
    df = df.copy()
    if "es_local" not in df.columns:
        df["es_local"] = 0
    for c in ["goles_favor", "goles_contra"] + features:
        df[c] = _a_numerico(df[c])
    base = df[["partido_id", "equipo"] + features]
    m = df.merge(base, on="partido_id", suffixes=("", "__rival"))
    m = m[m["equipo"] != m["equipo__rival"]].reset_index(drop=True)
    return m


def preparar_datos(df_raw):
    """
    Devuelve (d, features):
      d         -> DataFrame con features + resultado_victoria + partido_id
      features  -> lista de columnas predictoras finales
    """
    titulo("1. CARGA Y PREPARACIÓN DE DATOS")

    # --- 4.1 Detectar qué features existen realmente -------------------------
    if FORMATO_CSV == "ancho":
        disponibles = [f for f in FEATURES_BASE
                       if f"local_{f}" in df_raw.columns and f"visitante_{f}" in df_raw.columns]
    elif FORMATO_CSV == "largo":
        disponibles = [f for f in FEATURES_BASE if f in df_raw.columns]
    else:
        raise ValueError("FORMATO_CSV debe ser 'ancho' o 'largo'.")

    faltan = [f for f in FEATURES_BASE if f not in disponibles]
    if faltan:
        print(f"   Aviso: no se encontraron estas estadísticas y se omiten: {faltan}")
    if len(disponibles) < 2:
        raise ValueError("Se necesitan al menos 2 estadísticas. Revisá RENOMBRAR_COLUMNAS.")

    # --- 4.2 Pasar a una fila por equipo-partido -----------------------------
    d = (_ancho_a_largo(df_raw, disponibles) if FORMATO_CSV == "ancho"
         else _largo_con_rival(df_raw, disponibles))

    # --- 4.3 Variable objetivo ----------------------------------------------
    # 1 = el equipo ganó | 0 = empató o perdió
    d = d.dropna(subset=["goles_favor", "goles_contra"]).copy()
    d[OBJETIVO] = (d["goles_favor"] > d["goles_contra"]).astype(int)

    # --- 4.4 Features diferenciales (propio - rival) -------------------------
    if USAR_DIFERENCIAL:
        for f in disponibles:
            d[f] = d[f] - d[f"{f}__rival"]
    features = list(disponibles)
    if INCLUIR_LOCALIA:
        features.append("es_local")

    # --- 4.5 Calidad de datos ------------------------------------------------
    pct_nan = d[features].isna().mean()
    descartar = pct_nan[pct_nan > MAX_NAN_FEATURE].index.tolist()
    if descartar:
        print(f"   Aviso: se descartan por exceso de faltantes (> {MAX_NAN_FEATURE:.0%}): {descartar}")
        features = [f for f in features if f not in descartar]
    constantes = [f for f in features if d[f].nunique(dropna=True) <= 1]
    if constantes:
        print(f"   Aviso: se descartan por ser constantes: {constantes}")
        features = [f for f in features if f not in constantes]

    d = d[features + [OBJETIVO, "partido_id"]].reset_index(drop=True)
    d["partido_id"] = pd.factorize(d["partido_id"])[0]   # IDs enteros 0..G-1

    n_part = d["partido_id"].nunique()
    print(f"   Partidos: {n_part} | Observaciones (equipo-partido): {len(d)}")
    print(f"   Tasa base de victoria: {d[OBJETIVO].mean():.1%} "
          f"(el resto son empates o derrotas)")
    print(f"   Features usadas ({len(features)}): {features}")
    print(f"   Faltantes por feature (%):\n"
          f"{(d[features].isna().mean() * 100).round(2).to_string()}")
    print("   -> La imputación (mediana) se ajusta SOLO con el set de entrenamiento.")
    return d, features
```

```python
# 5. ANÁLISIS EXPLORATORIO (EDA)
# =============================================================================
def eda_heatmap(d, features):
    """Matriz de correlación entre métricas y victoria."""
    cols = features + [OBJETIVO]
    corr = d[cols].corr()
    mascara = np.triu(np.ones_like(corr, dtype=bool), k=1)
    etiquetas = [nombre_legible(c) for c in features] + ["VICTORIA"]

    fig, ax = plt.subplots(figsize=(10, 8.5))
    sns.heatmap(corr, mask=mascara, annot=True, fmt=".2f", cmap="RdBu_r",
                center=0, vmin=-1, vmax=1, square=True, linewidths=0.5,
                cbar_kws={"label": "Correlación de Pearson"}, ax=ax)
    ax.set_xticklabels(etiquetas, rotation=45, ha="right")
    ax.set_yticklabels(etiquetas, rotation=0)
    ax.set_title("Matriz de correlación entre las métricas del partido y la victoria")
    fig.tight_layout()
    guardar_figura(fig, "01_heatmap_correlacion")

    corr_obj = corr[OBJETIVO].drop(OBJETIVO)
    return corr_obj.reindex(corr_obj.abs().sort_values(ascending=False).index)


def eda_comparativos(d, corr_obj, k=6):
    """Boxplots y KDE: victoria vs. no victoria para las k métricas más correlacionadas."""
    top = [f for f in corr_obj.index if f != "es_local"][:k]
    if not top:
        return
    g = d.copy()
    g["grupo"] = np.where(g[OBJETIVO] == 1, ORDEN_GRUPOS[0], ORDEN_GRUPOS[1])
    sufijo = " (dif. propio − rival)" if USAR_DIFERENCIAL else ""
    ncols = 3
    nrows = int(np.ceil(len(top) / ncols))

    # --- Boxplots ---
    fig, axes = plt.subplots(nrows, ncols, figsize=(5.2 * ncols, 4.2 * nrows))
    axes = np.atleast_1d(axes).ravel()
    for ax, f in zip(axes, top):
        sns.boxplot(data=g, x="grupo", y=f, hue="grupo", order=ORDEN_GRUPOS,
                    hue_order=ORDEN_GRUPOS, palette=PALETA, dodge=False,
                    showfliers=False, ax=ax)
        if ax.get_legend() is not None:
            ax.get_legend().remove()
        ax.set_title(nombre_legible(f))
        ax.set_xlabel("")
        ax.set_ylabel(nombre_legible(f) + sufijo)
        ax.set_xticks(range(2))
        ax.set_xticklabels(["Victoria", "No victoria"])
    for ax in axes[len(top):]:
        ax.axis("off")
    fig.suptitle("Distribución de las métricas clave: partidos ganados vs. no ganados",
                 fontsize=14, fontweight="bold", y=1.02)
    fig.tight_layout()
    guardar_figura(fig, "02_boxplots_ganados_vs_no_ganados")

    # --- KDE ---
    fig, axes = plt.subplots(nrows, ncols, figsize=(5.2 * ncols, 4.0 * nrows))
    axes = np.atleast_1d(axes).ravel()
    for ax, f in zip(axes, top):
        sns.kdeplot(data=g, x=f, hue="grupo", hue_order=ORDEN_GRUPOS,
                    palette=PALETA, fill=True, common_norm=False, alpha=0.35,
                    linewidth=1.6, ax=ax)
        ax.set_title(nombre_legible(f))
        ax.set_xlabel(nombre_legible(f) + sufijo)
        ax.set_ylabel("Densidad")
        if ax.get_legend() is not None:
            ax.get_legend().set_title("")
    for ax in axes[len(top):]:
        ax.axis("off")
    fig.suptitle("Densidades (KDE) de las métricas clave según el resultado",
                 fontsize=14, fontweight="bold", y=1.02)
    fig.tight_layout()
    guardar_figura(fig, "03_kde_ganados_vs_no_ganados")
```

```python
# =============================================================================
# 6. MODELO 1 — REGRESIÓN LOGÍSTICA (interpretación: odds ratios)
#    Implementación propia (IRLS) con errores estándar robustos por clúster.
# =============================================================================
def ajustar_logit(X, y, grupos, max_iter=100, tol=1e-8):
    """
    Regresión logística por máxima verosimilitud (Newton-Raphson / IRLS).
    Covarianza 'sandwich' agrupada por partido (cluster-robust, con la
    corrección de muestra pequeña habitual).
    X: DataFrame (idealmente estandarizado) | y: Series 0/1 | grupos: array int
    """
    Xc = np.column_stack([np.ones(len(X)), X.values.astype(float)])
    yv = y.values.astype(float)
    n, k = Xc.shape
    beta = np.zeros(k)
    for _ in range(max_iter):
        eta = np.clip(Xc @ beta, -30, 30)
        p = 1 / (1 + np.exp(-eta))
        W = np.clip(p * (1 - p), 1e-9, None)
        H = Xc.T @ (Xc * W[:, None]) + 1e-10 * np.eye(k)
        paso = np.linalg.solve(H, Xc.T @ (yv - p))
        beta = beta + paso
        if np.max(np.abs(paso)) < tol:
            break

    eta = np.clip(Xc @ beta, -30, 30)
    p = 1 / (1 + np.exp(-eta))
    W = np.clip(p * (1 - p), 1e-9, None)
    H_inv = np.linalg.inv(Xc.T @ (Xc * W[:, None]) + 1e-10 * np.eye(k))
    scores = Xc * (yv - p)[:, None]
    sumas = pd.DataFrame(scores).groupby(np.asarray(grupos)).sum().values
    G = sumas.shape[0]
    correccion = (G / (G - 1)) * ((n - 1) / (n - k))
    V = correccion * H_inv @ (sumas.T @ sumas) @ H_inv
    se = np.sqrt(np.diag(V))
    z = beta / se
    nombres = ["const"] + list(X.columns)
    return {
        "nombres": nombres,
        "beta": pd.Series(beta, index=nombres),
        "se": pd.Series(se, index=nombres),
        "pvalor": pd.Series(2 * norm.sf(np.abs(z)), index=nombres),
        "ic_inf": pd.Series(beta - 1.96 * se, index=nombres),
        "ic_sup": pd.Series(beta + 1.96 * se, index=nombres),
    }


def predecir_logit(modelo, X):
    eta = modelo["beta"]["const"] + X.values @ modelo["beta"].drop("const").values
    return 1 / (1 + np.exp(-np.clip(eta, -30, 30)))


def calcular_vif(X):
    """VIF = diagonal de la inversa de la matriz de correlación."""
    corr = np.corrcoef(X.values, rowvar=False)
    return pd.Series(np.diag(np.linalg.inv(corr)), index=X.columns)


def tabla_odds_ratios(modelo, scaler, features):
    """Coeficientes y odds ratios por 1 desvío estándar y por unidad natural."""
    b = modelo["beta"][features]
    sd = pd.Series(scaler.scale_, index=features)
    t = pd.DataFrame({
        "coef_estandarizado": b,
        "OR_1SD": np.exp(b),
        "IC95_inf": np.exp(modelo["ic_inf"][features]),
        "IC95_sup": np.exp(modelo["ic_sup"][features]),
        "p_valor": modelo["pvalor"][features],
        "desvio_estandar": sd,
    })
    t["coef_por_unidad"] = t["coef_estandarizado"] / sd
    t["OR_por_unidad"] = np.exp(t["coef_por_unidad"])
    return t.sort_values("coef_estandarizado", key=np.abs, ascending=False)


def grafico_forest(tabla):
    """Forest plot de odds ratios (por 1 DE) con IC 95 %, escala logarítmica."""
    t = tabla.sort_values("OR_1SD")
    y = np.arange(len(t))
    colores = np.where(t["p_valor"] < 0.05, "#1b7837", "#7f7f7f")
    fig, ax = plt.subplots(figsize=(9, 0.55 * len(t) + 2.4))
    ax.errorbar(t["OR_1SD"], y,
                xerr=[t["OR_1SD"] - t["IC95_inf"], t["IC95_sup"] - t["OR_1SD"]],
                fmt="none", ecolor="#555555", capsize=3, lw=1.3, zorder=2)
    ax.scatter(t["OR_1SD"], y, c=colores, s=70, zorder=3)
    ax.axvline(1, ls="--", color="black", lw=1)
    ax.set_xscale("log")
    ax.xaxis.set_major_formatter(ScalarFormatter())
    ax.xaxis.set_minor_formatter(NullFormatter())
    ax.set_yticks(y)
    ax.set_yticklabels([nombre_legible(f) for f in t.index])
    ax.set_xlabel("Odds Ratio por +1 desvío estándar (escala log, IC 95 %)")
    ax.set_title("Regresión logística: impacto de cada métrica sobre las odds de ganar")
    ax.legend(handles=[Patch(color="#1b7837", label="p < 0.05"),
                       Patch(color="#7f7f7f", label="No significativo")],
              loc="lower right", frameon=True)
    fig.tight_layout()
    guardar_figura(fig, "04_forest_odds_ratios_logit")

```

```python
# 7. MODELO 2 — RANDOM FOREST (importancia, relaciones no lineales)
# =============================================================================
def entrenar_random_forest(X_train, y_train):
    rf = RandomForestClassifier(n_estimators=500, min_samples_leaf=5,
                                max_features="sqrt", n_jobs=-1,
                                random_state=SEMILLA)
    rf.fit(X_train, y_train)
    return rf


def grafico_importancias_rf(imp_mdi, imp_perm):
    """Importancia por impureza (MDI) vs. por permutación (más confiable)."""
    fig, axes = plt.subplots(1, 2, figsize=(14, 0.5 * len(imp_mdi) + 2.6))
    m = imp_mdi.sort_values()
    axes[0].barh([nombre_legible(f) for f in m.index], m.values, color="#4575b4")
    axes[0].set_title("Importancia por impureza (MDI)")
    axes[0].set_xlabel("Importancia (suma = 1)")

    p = imp_perm.sort_values("media")
    axes[1].barh([nombre_legible(f) for f in p.index], p["media"].values,
                 xerr=p["desvio"].values, color="#d6604d", capsize=3)
    axes[1].set_title("Importancia por permutación (test, ΔAUC)")
    axes[1].set_xlabel("Caída del ROC-AUC al permutar la variable")
    fig.suptitle("Random Forest: importancia relativa de las variables",
                 fontsize=14, fontweight="bold")
    fig.tight_layout()
    guardar_figura(fig, "05_importancias_random_forest")
```

```python
# =============================================================================
# 8. EVALUACIÓN
# =============================================================================
def evaluar(nombre, y_true, proba):
    pred = (proba >= 0.5).astype(int)
    return {"modelo": nombre,
            "accuracy": accuracy_score(y_true, pred),
            "roc_auc": roc_auc_score(y_true, proba),
            "cm": confusion_matrix(y_true, pred),
            "pred": pred, "proba": proba}


def graficos_evaluacion(resultados, y_test):
    # Matrices de confusión
    fig, axes = plt.subplots(1, len(resultados), figsize=(5.6 * len(resultados), 4.8))
    axes = np.atleast_1d(axes)
    for ax, r in zip(axes, resultados):
        sns.heatmap(r["cm"], annot=True, fmt="d", cmap="Blues", cbar=False,
                    square=True, linewidths=0.5, ax=ax,
                    xticklabels=["No victoria", "Victoria"],
                    yticklabels=["No victoria", "Victoria"])
        ax.set_title(f"{r['modelo']}\nAcc = {r['accuracy']:.3f} | AUC = {r['roc_auc']:.3f}")
        ax.set_xlabel("Predicción")
        ax.set_ylabel("Valor real")
    fig.suptitle("Matrices de confusión (set de test)", fontsize=14, fontweight="bold")
    fig.tight_layout()
    guardar_figura(fig, "06_matrices_confusion")

    # Curvas ROC
    fig, ax = plt.subplots(figsize=(6.5, 5.8))
    for r, c in zip(resultados, ["#2166ac", "#b2182b"]):
        fpr, tpr, _ = roc_curve(y_test, r["proba"])
        ax.plot(fpr, tpr, lw=2, color=c, label=f"{r['modelo']} (AUC = {r['roc_auc']:.3f})")
    ax.plot([0, 1], [0, 1], "k--", lw=1, label="Azar (AUC = 0.500)")
    ax.set_xlabel("Tasa de falsos positivos")
    ax.set_ylabel("Tasa de verdaderos positivos")
    ax.set_title("Curvas ROC en el set de test")
    ax.legend(loc="lower right")
    fig.tight_layout()
    guardar_figura(fig, "07_curvas_roc")


def validacion_cruzada(X, y, grupos):
    """AUC con GroupKFold (5 pliegues) para chequear estabilidad del desempeño."""
    cv = GroupKFold(n_splits=5)
    pipes = {
        "Regresión Logística": make_pipeline(SimpleImputer(strategy="median"), StandardScaler(),
                                             LogisticRegression(C=1e6, max_iter=5000)),
        "Random Forest": make_pipeline(SimpleImputer(strategy="median"),
                                       RandomForestClassifier(n_estimators=300, min_samples_leaf=5,
                                                              n_jobs=-1, random_state=SEMILLA)),
    }
    out = {}
    for nombre, pipe in pipes.items():
        s = cross_val_score(pipe, X, y, groups=grupos, cv=cv, scoring="roc_auc")
        out[nombre] = (s.mean(), s.std())
    return out
```

```python
# =============================================================================
# 9. RANKING COMBINADO Y GRÁFICO FINAL (Top 10)
# =============================================================================
def construir_ranking(tabla_or, imp_perm):
    r = pd.DataFrame({
        "importancia_logit": tabla_or["coef_estandarizado"].abs(),
        "efecto": np.sign(tabla_or["coef_estandarizado"]),
        "importancia_rf": imp_perm["media"].clip(lower=0),
    })
    for c in ("importancia_logit", "importancia_rf"):
        s = r[c].sum()
        r[c + "_norm"] = r[c] / s if s > 0 else 0.0
    r["score_combinado"] = r[["importancia_logit_norm", "importancia_rf_norm"]].mean(axis=1)
    return r.sort_values("score_combinado", ascending=False)


def grafico_top10(ranking, n=10):
    top = ranking.head(n).iloc[::-1]  # invertido para que el #1 quede arriba
    colores = np.where(top["efecto"] > 0, "#1b7837", "#b2182b")
    fig, ax = plt.subplots(figsize=(9.5, 0.55 * len(top) + 2.4))
    barras = ax.barh([nombre_legible(f) for f in top.index],
                     top["score_combinado"], color=colores, edgecolor="black", lw=0.6)
    for b, v in zip(barras, top["score_combinado"]):
        ax.text(v + 0.003, b.get_y() + b.get_height() / 2, f"{v:.3f}",
                va="center", fontsize=9)
    ax.set_xlim(0, top["score_combinado"].max() * 1.15)
    ax.set_xlabel("Puntaje combinado de importancia (promedio normalizado: |coef. logit| y permutación RF)")
    ax.set_ylabel("Estadística")
    ax.set_title(f"Top {len(top)} estadísticas más determinantes para ganar un partido")
    ax.legend(handles=[Patch(color="#1b7837", label="Efecto positivo sobre la victoria"),
                       Patch(color="#b2182b", label="Efecto negativo sobre la victoria")],
              loc="lower right", frameon=True)
    fig.tight_layout()
    guardar_figura(fig, "08_top10_estadisticas_determinantes")
```

```python
# =============================================================================
# 10. INTERPRETACIÓN EN CONSOLA (en español)
# =============================================================================
def imprimir_interpretacion(tabla_or, ranking, resultados, cv, tasa_base, vif, n_top=10):
    titulo("RESULTADOS E INTERPRETACIÓN")
    lg, rf = resultados

    print("\n▶ DESEMPEÑO PREDICTIVO (set de test, 20 % de los partidos)")
    print(f"   • Regresión Logística : Accuracy = {lg['accuracy']:.3f} | ROC-AUC = {lg['roc_auc']:.3f}")
    print(f"   • Random Forest       : Accuracy = {rf['accuracy']:.3f} | ROC-AUC = {rf['roc_auc']:.3f}")
    print(f"   • Referencia: predecir siempre 'no victoria' daría ≈ {1 - tasa_base:.3f} de accuracy.")
    print("   • Validación cruzada agrupada (AUC media ± DE, 5 pliegues):")
    for n, (m, s) in cv.items():
        print(f"       - {n}: {m:.3f} ± {s:.3f}")
    print("   Lectura: un AUC de 0.5 es azar y 1.0 es perfección; entre 0.70 y 0.80 se\n"
          "   considera un desempeño aceptable-bueno para un fenómeno tan aleatorio como el fútbol.")

    print("\n▶ REGRESIÓN LOGÍSTICA — efecto de cada métrica (resto de variables constante)")
    sufijo = " de ventaja sobre el rival" if USAR_DIFERENCIAL else ""
    p0 = tasa_base
    odds0 = p0 / (1 - p0)
    orden = ranking.head(n_top).index
    for f in orden:
        fila = tabla_or.loc[f]
        nombre, unidad, paso = info_etiqueta(f)
        sig = "estadísticamente significativo" if fila["p_valor"] < 0.05 else "NO significativo al 5 %"
        if f == "es_local":
            or_p = fila["OR_por_unidad"]
            lo, hi = fila["IC95_inf"] ** (1 / fila["desvio_estandar"]), fila["IC95_sup"] ** (1 / fila["desvio_estandar"])
            frase = "jugar de local (vs. visitante)"
        else:
            k = paso / fila["desvio_estandar"]
            or_p = np.exp(fila["coef_por_unidad"] * paso)
            lo, hi = fila["IC95_inf"] ** k, fila["IC95_sup"] ** k
            frase = f"cada +{paso:g} {unidad}{sufijo}"
        pct = (or_p - 1) * 100
        verbo = "aumentan" if pct > 0 else "disminuyen"
        p_nueva = odds0 * or_p / (1 + odds0 * or_p)
        print(f"   • {nombre}: por {frase}, las probabilidades (odds) de ganar {verbo} "
              f"en {abs(pct):.1f} %\n"
              f"       OR = {or_p:.3f} (IC95 % {min(lo, hi):.3f}–{max(lo, hi):.3f}); {sig} "
              f"({fmt_p(fila['p_valor'])}).\n"
              f"       Desde la tasa base de victoria ({p0:.1%}), equivale a pasar a ≈ {p_nueva:.1%}.")

    print("\n▶ RANKING COMBINADO (promedio de la importancia en ambos modelos)")
    vista = ranking.head(n_top)[["score_combinado", "importancia_logit_norm", "importancia_rf_norm", "efecto"]].copy()
    vista.index = [nombre_legible(f) for f in vista.index]
    vista["efecto"] = np.where(vista["efecto"] > 0, "+", "−")
    vista.columns = ["Puntaje", "Peso logit", "Peso RF", "Signo"]
    print(vista.round(3).to_string())

    rho = ranking[["importancia_logit", "importancia_rf"]].corr(method="spearman").iloc[0, 1]
    print(f"\n   Concordancia entre modelos (Spearman de los rankings): {rho:.2f}")
    print("   (cercana a 1 => ambos modelos coinciden en qué variables importan)")

    mejores = [nombre_legible(f) for f in ranking.index[:3]]
    signif = [nombre_legible(f) for f in ranking.index if tabla_or.loc[f, "p_valor"] < 0.05]
    print("\n▶ CONCLUSIÓN")
    print(f"   Top 3 por importancia combinada: {', '.join(mejores)}.")
    if signif:
        print(f"   Con efecto estadísticamente significativo (p < 0.05) en la regresión logística: "
              f"{', '.join(signif)}.")
    else:
        print("   Ninguna métrica resultó significativa al 5 % en la regresión logística.")

    altos = vif[vif > 5]
    print("\n▶ ADVERTENCIAS METODOLÓGICAS")
    if len(altos):
        txt = ", ".join(f"{nombre_legible(k)} (VIF = {v:.1f})" for k, v in altos.items())
        print(f"   • Multicolinealidad alta (VIF > 5): {txt}.\n"
              f"     Sus coeficientes logit pueden ser inestables y la importancia del Random Forest\n"
              f"     puede repartirse entre variables correlacionadas; interpretalas en conjunto.")
    else:
        print("   • No se detecta multicolinealidad preocupante (todos los VIF ≤ 5).")
    print("   • Asociación ≠ causalidad: ganar también hace que el rival se tire atrás, que el\n"
          "     ganador administre la posesión, etc. Las métricas reflejan en parte el estado del juego.\n"
          "   • Las odds NO son probabilidades: un OR de 1.30 significa +30 % en las odds, no +30 puntos\n"
          "     de probabilidad (por eso se muestra también la traducción a probabilidad).\n"
          "   • 'No victoria' mezcla empates y derrotas; el modelo mide 'ganar vs. no ganar'.")
```

```python
# =============================================================================
# 11. PIPELINE PRINCIPAL
# =============================================================================
def main():
    configurar_estilo()

    # --- Datos ---
    df_raw = cargar_datos()
    d, features = preparar_datos(df_raw)

    # --- EDA ---
    titulo("2. ANÁLISIS EXPLORATORIO (EDA)")
    corr_obj = eda_heatmap(d, features)
    print("\n   Correlación de cada métrica con la victoria (ordenada por magnitud):")
    print(corr_obj.rename(index=nombre_legible).round(3).to_string())
    eda_comparativos(d, corr_obj)

    # --- Split 80/20 agrupado por partido ---
    titulo("3. MODELADO")
    X, y, grupos = d[features], d[OBJETIVO], d["partido_id"].values
    gss = GroupShuffleSplit(n_splits=1, test_size=TEST_SIZE, random_state=SEMILLA)
    idx_tr, idx_te = next(gss.split(X, y, groups=grupos))
    X_tr, X_te = X.iloc[idx_tr], X.iloc[idx_te]
    y_tr, y_te = y.iloc[idx_tr], y.iloc[idx_te]
    g_tr = grupos[idx_tr]
    print(f"   Train: {len(X_tr)} obs. ({len(set(g_tr))} partidos) | "
          f"Test: {len(X_te)} obs. ({len(set(grupos[idx_te]))} partidos)")

    # Imputación + escalado ajustados SOLO con train
    imputador = SimpleImputer(strategy="median").fit(X_tr)
    X_tr_i = pd.DataFrame(imputador.transform(X_tr), columns=features, index=X_tr.index)
    X_te_i = pd.DataFrame(imputador.transform(X_te), columns=features, index=X_te.index)
    scaler = StandardScaler().fit(X_tr_i)
    X_tr_s = pd.DataFrame(scaler.transform(X_tr_i), columns=features, index=X_tr.index)
    X_te_s = pd.DataFrame(scaler.transform(X_te_i), columns=features, index=X_te.index)

    # --- Modelo 1: Regresión logística ---
    vif = calcular_vif(X_tr_s)
    modelo_lg = ajustar_logit(X_tr_s, y_tr, g_tr)
    tabla_or = tabla_odds_ratios(modelo_lg, scaler, features)
    proba_lg = predecir_logit(modelo_lg, X_te_s)
    print("\n   Modelo 1 (Regresión Logística) — tabla de coeficientes y odds ratios:")
    mostrar = tabla_or.rename(index=nombre_legible)[
        ["coef_estandarizado", "OR_1SD", "IC95_inf", "IC95_sup", "p_valor", "OR_por_unidad"]]
    print(mostrar.round(4).to_string())
    grafico_forest(tabla_or)

    # --- Modelo 2: Random Forest ---
    rf = entrenar_random_forest(X_tr_i, y_tr)
    proba_rf = rf.predict_proba(X_te_i)[:, 1]
    imp_mdi = pd.Series(rf.feature_importances_, index=features)
    perm = permutation_importance(rf, X_te_i, y_te, n_repeats=30, scoring="roc_auc",
                                  random_state=SEMILLA, n_jobs=-1)
    imp_perm = pd.DataFrame({"media": perm.importances_mean, "desvio": perm.importances_std},
                            index=features)
    grafico_importancias_rf(imp_mdi, imp_perm)

    # --- Evaluación ---
    titulo("4. EVALUACIÓN")
    res_lg = evaluar("Regresión Logística", y_te, proba_lg)
    res_rf = evaluar("Random Forest", y_te, proba_rf)
    for r in (res_lg, res_rf):
        print(f"\n   {r['modelo']}: Accuracy = {r['accuracy']:.3f} | ROC-AUC = {r['roc_auc']:.3f}")
        print(classification_report(y_te, r["pred"], target_names=["No victoria", "Victoria"], digits=3))
    graficos_evaluacion([res_lg, res_rf], y_te)
    cv = validacion_cruzada(X, y, grupos)

    # --- Ranking final ---
    ranking = construir_ranking(tabla_or, imp_perm)
    grafico_top10(ranking)

    # --- Guardar tablas (para LaTeX / informe) ---
    tabla_or.to_csv(os.path.join(DIR_SALIDA, "tabla_odds_ratios.csv"), encoding="utf-8-sig")
    ranking.to_csv(os.path.join(DIR_SALIDA, "ranking_importancia.csv"), encoding="utf-8-sig")
    pd.DataFrame([{k: r[k] for k in ("modelo", "accuracy", "roc_auc")} for r in (res_lg, res_rf)]
                 ).to_csv(os.path.join(DIR_SALIDA, "metricas_modelos.csv"), index=False, encoding="utf-8-sig")
    print(f"\n   [tablas guardadas] {DIR_SALIDA}/tabla_odds_ratios.csv, ranking_importancia.csv, metricas_modelos.csv")

    # --- Interpretación final ---
    imprimir_interpretacion(tabla_or, ranking, [res_lg, res_rf], cv, y.mean(), vif)

    # --- Descarga (Colab) ---
    if DESCARGAR_ZIP and en_colab():
        from google.colab import files
        shutil.make_archive(DIR_SALIDA, "zip", DIR_SALIDA)
        files.download(DIR_SALIDA + ".zip")


if __name__ == "__main__":
    main()
```
